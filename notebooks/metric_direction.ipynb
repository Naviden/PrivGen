```python
import inspect
import synthcity.metrics.eval_privacy as eval_privacy

# Iterate through all members of the eval_privacy module
for name, obj in inspect.getmembers(eval_privacy):
    # Check if the member is a class and has a 'direction' method
    if inspect.isclass(obj) and hasattr(obj, 'direction'):
        # Call the 'direction' method and print the result
        print(f"{name}.direction(): {obj.direction()}")
```
